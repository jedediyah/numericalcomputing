# Regression

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

Below is an example of linear regression using NumPy's `polyfit` function. This function takes an argument that is the degree of the polynomial you want to fit. So you can change this to a cubic regression by change the 1 to a 3.

````python
# An example regression.
x = np.array([1, 2, 3, 4, 5])
y = np.array([4, 3, 4, 6, 9])

coeff = np.polyfit(x, y, 1)

print(coeff)
````

Those coefficients are the coefficients of your polynomial. You can plot your best fit curve to your data with the following.

````python
# Plot the data
plt.scatter(x,y)

# Plot the regression
x_fit = np.linspace(1,5,50)
y_fit = coeff[0]*x_fit + coeff[1]
plt.plot( x_fit,y_fit )
plt.show()
````
